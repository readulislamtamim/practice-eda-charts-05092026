```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
df.drop(columns=["Cabin"], inplace=True)
```

```python
df.isnull().sum()
```

```python
sns.countplot(x = "Survived", data = df)
plt.title("Survival Count")
plt.show()
```

```python
counts = df["Survived"].value_counts()
x = counts.index
y = counts.values
plt.bar(x, y)
plt.title("Survival Count")
plt.xticks([1, 0], labels=["Survived", "Died"])
plt.xlabel("Survived")
plt.ylabel("Count")
plt.show()
```

```python
sns.countplot(x = "Sex", data=df)
plt.title("Gender distribution")
plt.show()
```

```python
counts = df["Sex"].value_counts()
x = counts.index
y = counts.values

plt.bar(x, y)
plt.title("Gender distribution")
plt.show()
```

```python
sns.histplot(df["Age"],
             bins = 10,
             kde = True)
plt.title("Age Distribution")
plt.show()
```

```python
mean_age = df["Age"].mean()
plt.hist(df["Age"], bins = 10, color = "coral", edgecolor = "b")
plt.axvline(mean_age)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
counts = df["Survived"].value_counts()
explode = [.1, 0]

plt.pie(counts, 
        labels=["Survived", "Died"], 
        autopct="%1.1f%%",
        explode=explode)
plt.title("Survival Percentage")
plt.show()
```

```python
x = df["Fare"]
sns.boxplot(x)
plt.show()
```

```python
df = df[df["Fare"] < 55]
```

```python
x = df["Fare"]
sns.boxplot(x)
plt.show()
```

```python
sns.scatterplot(x = "Age",
                y = "Fare",
                hue="Sex",
                data= df)
plt.title("Age vs Fare")
plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True),
            annot=True)
plt.show()
```
